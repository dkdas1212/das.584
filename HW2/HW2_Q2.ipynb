```python
def sigmoid(v):
    """
    Sigmoid activation function.

    Input:
        v : net input (scalar or numpy array)

    Output:
        y : sigmoid(v)
    """
    
    # TODO (students): implement sigmoid activation
    y = 1 / (1 + np.exp(-v))
    return y
```
